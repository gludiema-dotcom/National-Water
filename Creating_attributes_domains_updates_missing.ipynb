```python
import arcpy
import pandas as pd
import os
```

```python
# ==========================
# User Inputs
# ==========================

gdb = r"C:\National Water Utility\test.gdb"
excel_file = r"C:\National Water Utility\Domains.xlsx"
sheet_name = "Domains"
```

```python
# ==========================
# Read Excel
# ==========================

df = pd.read_excel(excel_file, sheet_name=sheet_name)

# Remove blank rows
df = df.dropna(subset=["DomainName", "Code"])

# Get existing domains
existing_domains = {d.name: d for d in arcpy.da.ListDomains(gdb)}

# ==========================
# Process each domain
# ==========================

for domain_name in df["DomainName"].unique():

    domain_rows = df[df["DomainName"] == domain_name]

    field_type = str(domain_rows.iloc[0]["FieldType"]).upper()

    #------------------------------------
    # Create domain if it doesn't exist
    #------------------------------------
    if domain_name not in existing_domains:

        arcpy.management.CreateDomain(
            in_workspace=gdb,
            domain_name=domain_name,
            domain_description=domain_name,
            field_type=field_type,
            domain_type="CODED"
        )

        print(f"Created Domain: {domain_name}")

        # Refresh domain list
        existing_domains = {d.name: d for d in arcpy.da.ListDomains(gdb)}

    #------------------------------------
    # Existing coded values
    #------------------------------------
    existing_codes = existing_domains[domain_name].codedValues

    #------------------------------------
    # Add only missing codes
    #------------------------------------
    for _, row in domain_rows.iterrows():

        code = row["Code"]
        description = row["Description"]

        if code not in existing_codes:

            arcpy.management.AddCodedValueToDomain(
                in_workspace=gdb,
                domain_name=domain_name,
                code=code,
                code_description=description
            )

            print(f"Added {code} -> {description} to {domain_name}")

        else:
            print(f"{domain_name}: Code '{code}' already exists.")

print("\nFinished Creating Attributes Domains.")
```
